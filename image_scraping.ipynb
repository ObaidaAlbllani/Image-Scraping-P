```python
from selenium import webdriver
from bs4 import BeautifulSoup
import requests
import time

#  Any infinity scroll URL
var = "trees"
url = "https://in.pinterest.com/search/pins/?q=" + var 
ScrollNumber = 4  # The depth we wish to load
sleepTimer = 1    # Waiting 1 second for page to load
resolvedURLs = []


options = webdriver.ChromeOptions() 
options.add_experimental_option("excludeSwitches", ["enable-logging"])

driver = webdriver.Chrome(options=options)  
driver.get(url)

for _ in range(1,ScrollNumber):
    driver.execute_script("window.scrollTo(1,100000)")
    print("scrolling")
    time.sleep(sleepTimer)

soup = BeautifulSoup(driver.page_source,'html.parser')

driver.close()

```

```python
for link in soup.find_all('img'):
   src = link.get('src')
   resolvedURLs.append(requests.compat.urljoin(url, src))
   #print(link.get('alt'))


print(resolvedURLs)

for re_url in range(len(resolvedURLs)):
    resolvedURLs[re_url]=resolvedURLs[re_url].replace("236x", "originals")

print(resolvedURLs)




for image in resolvedURLs:
    webs = requests.get(image)
    open('C:/Users/ASUS/GP/test_img/' + image.split('/')[-1], 'wb').write(webs.content)
   
```
